## 🧭 Phase exploratoire — V1 (10/07 → 10/08/2026)

**Chapitres traités : §1 à §5** _(consignés dans une première version — registre ADR, notes de travail et un premier notebook — pas encore dans le notebook V2)_

Première version du projet. Les questions des chapitres §1 à §5 y ont été travaillées — cadrage métier, données, éthique, compréhension des données — mais dans des documents à part et un premier notebook. Beaucoup de tâtonnement, notamment sur le cadre méthodologique et la gouvernance.

### 🔧 Sujets travaillés
- **10/07** — initialisation du dépôt
- **27/07** — premières fondations du projet
- **07/08** — mise en place de la stack ML, du `README` et du `CLAUDE.md`, des agents Claude Code spécialisés, et d'un registre de décisions au format ADR
- **09/08** — cadrage RGPD / éthique consigné en **12 fiches ADR** : non-versionnement des données, gouvernance de l'assistance IA, base légale (mission d'intérêt public), périmètre de responsabilité, article 22 et supervision humaine, équité (sexe/boursier hors entraînement), fuites inter-cibles et temporelle (mi-S1), variables leurres, explicabilité SHAP — matière reprise dans les chapitres §2, §3 et §4 de la V2
- **10/08** — outillage de profilage, **rapports de profil sur les deux jeux** (catalogue + étudiants), premier palier de données (bronze) et compréhension des données (§5)

### 📝 Difficultés
- **Travail avec l'IA difficile à cadrer.** Les échanges libres partaient dans tous les sens et manquaient de contrôle. → Mise en place d'**agents spécialisés et délimités** (un rôle par agent : méthode ML, RGPD, couverture du cas d'usage), chacun en lecture seule, pour cadrer et challenger plutôt que produire.
- **Registre ADR trop verbeux.** Le format ADR, très détaillé, a vite posé des **problèmes de cohérence entre fiches** (décisions qui se recoupaient ou se contredisaient). → Décision de **repartir sur une base simplifiée (V2)** : intégrer les décisions **directement dans le notebook** (journal de bord) et n'en garder qu'un **registre CSV léger** pour la navigation.
